```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import yfinance as yf

# ==============================
# 1. ATIVOS (BANCOS)
# ==============================
ativos = ["ITUB4.SA", "BBDC4.SA", "BBAS3.SA", "SANB11.SA"]

# ==============================
# 2. BAIXAR DADOS REAIS
# ==============================
dados = yf.download(ativos, start="2020-01-01")['Close']

# ==============================
# 3. TRANSFORMAR EM MENSAL
# ==============================
dados_mensal = dados.resample('M').last()

# ==============================
# 4. CALCULAR RETORNOS MENSAIS
# ==============================
retornos = dados_mensal.pct_change().dropna()

# ==============================
# 5. CONFIGURAÇÃO DO INVESTIMENTO
# ==============================
capital_inicial = 100000
aporte_mensal = 1000

# pesos iniciais (igual)
pesos = np.array([0.25, 0.25, 0.25, 0.25])

# ==============================
# 6. SIMULAÇÃO COM APORTES + REBALANCEAMENTO
# ==============================
valores = [capital_inicial]
pesos_atuais = pesos.copy()

for i in range(len(retornos)):

    # retorno do mês
    r = retornos.iloc[i]

    # valor atual da carteira
    valor_atual = valores[-1]

    # aplica retorno
    valor_atual = valor_atual * (1 + np.dot(r, pesos_atuais))

    # aporte mensal
    valor_atual += aporte_mensal

    # rebalanceamento (volta pros pesos originais todo mês)
    pesos_atuais = pesos.copy()

    valores.append(valor_atual)

# ==============================
# 7. COMPARAÇÃO INDIVIDUAL
# ==============================
capital_base = capital_inicial

valores_individuais = {}
for ativo in ativos:
    serie = dados_mensal[ativo]
    valores_individuais[ativo] = (serie / serie.iloc[0]) * capital_base

# ==============================
# 8. GRÁFICO INTERATIVO
# ==============================
fig = go.Figure()

# ativos individuais
for ativo in ativos:
    fig.add_trace(go.Scatter(
        x=dados_mensal.index,
        y=valores_individuais[ativo],
        mode='lines',
        name=ativo
    ))

# carteira
fig.add_trace(go.Scatter(
    x=dados_mensal.index.insert(0, dados_mensal.index[0]),
    y=valores,
    mode='lines',
    name='Carteira Bancos',
    line=dict(width=4)
))

fig.update_layout(
    title='Simulação Mensal - Investimento em Bancos Brasileiros',
    xaxis_title='Tempo (Mês/Ano)',
    yaxis_title='Valor (R$)',
    xaxis=dict(tickformat="%m/%Y"),
    template='plotly_dark',

    plot_bgcolor='black',
    paper_bgcolor='black',

    font=dict(color='white'),

    legend=dict(
        bgcolor='rgba(0,0,0,0)'
    )
)

fig.show()

# ==============================
# 9. RESULTADO FINAL
# ==============================
print(f"Valor final da carteira: R$ {valores[-1]:.2f}")
print(f"Total investido: R$ {capital_inicial + aporte_mensal * len(retornos):.2f}")
print(f"Lucro: R$ {valores[-1] - (capital_inicial + aporte_mensal * len(retornos)):.2f}")
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go

# ==============================
# ENTRADA
# ==============================
ativos_input = input("Digite os ativos (ex: ITUB4.SA, BBDC4.SA, BBAS3.SA, SANB11.SA): ")
ativos = [a.strip().upper() for a in ativos_input.split(",")]

capital_inicial = 10000
aporte_mensal = 1000

# ==============================
# DADOS
# ==============================
dados = yf.download(ativos, start="2020-01-01")['Close']

# transformar em mensal
dados_mensal = dados.resample('M').last()

# ==============================
# RETORNOS
# ==============================
retornos = dados_mensal.pct_change().dropna()

# ==============================
# MÉTRICAS
# ==============================
retorno_acumulado = (1 + retornos).prod() - 1
volatilidade = retornos.std() * np.sqrt(12)
correlacao = retornos.corr()

# melhor ativo
melhor_ativo = retorno_acumulado.idxmax()

# ==============================
# SIMULAÇÃO (APORTES)
# ==============================
pesos = np.ones(len(ativos)) / len(ativos)
valores = [capital_inicial]

for i in range(len(retornos)):
    r = retornos.iloc[i]
    valor = valores[-1]

    valor = valor * (1 + np.dot(r, pesos))
    valor += aporte_mensal

    valores.append(valor)

# ==============================
# RESULTADOS
# ==============================
print("\n📊 RESULTADOS\n")
print(f"🏆 Melhor ativo: {melhor_ativo}\n")

print("📈 Retorno acumulado:")
print((retorno_acumulado * 100).round(2).astype(str) + "%\n")

print("⚠️ Volatilidade:")
print((volatilidade * 100).round(2).astype(str) + "%\n")

print("🔗 Correlação:")
print(correlacao.round(2))

# ==============================
# GRÁFICO INTERATIVO
# ==============================
fig = go.Figure()

# ativos individuais
for ativo in ativos:
    fig.add_trace(go.Scatter(
        x=dados_mensal.index,
        y=(dados_mensal[ativo] / dados_mensal[ativo].iloc[0]) * capital_inicial,
        mode='lines',
        name=ativo
    ))

# carteira
fig.add_trace(go.Scatter(
    x=dados_mensal.index.insert(0, dados_mensal.index[0]),
    y=valores,
    mode='lines',
    name='Carteira',
    line=dict(width=4)
))

fig.update_layout(
    template='plotly_dark',
    title='Simulação Mensal - Investimento em Bancos',
    xaxis_title='Tempo (Mês/Ano)',
    yaxis_title='Valor (R$)',
    xaxis=dict(tickformat="%m/%Y")
)

fig.show()
```

```python
import os

print('*******************************')
print('*** Bem-vindo ao jogo da Forca ***')
print('*******************************')

palavra_secreta = input('Digite uma palavra: ').lower()
os.system('cls' if os.name == 'nt' else 'clear')  # limpa tela

letras_acertadas = ['_' for _ in palavra_secreta]
letras_tentadas = set()

erros = 0
max_erros = 6

while True:
    print("\nPalavra:", " ".join(letras_acertadas))
    print("Tentativas:", ", ".join(sorted(letras_tentadas)))

    chute = input("Digite uma letra: ").lower().strip()

    if len(chute) != 1 or not chute.isalpha():
        print("Digite apenas UMA letra válida.")
        continue

    if chute in letras_tentadas:
        print("Você já tentou essa letra.")
        continue

    letras_tentadas.add(chute)

    if chute in palavra_secreta:
        for i, letra in enumerate(palavra_secreta):
            if letra == chute:
                letras_acertadas[i] = letra
    else:
        erros += 1
        print(f"Errou! Restam {max_erros - erros} tentativas.")

    if '_' not in letras_acertadas:
        print("\nParabéns! Você acertou:", palavra_secreta)
        break

    if erros >= max_erros:
        print("\nVocê perdeu! A palavra era:", palavra_secreta)
        break
```

```python
ARQUIVO = "cadastro.txt"
def adicionar():
  nome = input("Nome: ")
  idade = input("Idade: ")
  with open(ARQUIVO, "A") as f:
    f.write(f"{nome},{idade}\n")

def listar():
  try:
      with open(ARQUIVO, "r") as f:
        for i, linha in enumerate(f):
          nome, idade = liha.strip().split(",")
          print(f"{i} - Nome: {nome} | Idade: {idade}")
  except FileNotFoundError:
    print("Arquivo não encontrado.")

def buscar():
  nome_busca = input("DIgite o nome:")
  with open(ARQUIVO, "r") as f:
    for i, linha in f:
      nome, idade = linha.strip().split(",")
      if nome_busca.lower() in nome.lower():
        print(f"Encontrado: {nome}- {idade}")

def atualizar():
  listar()
  indice = int(input("Digite o indice: "))
  with open(ARQUIVO, "r") as f:
      linhas = f.readlines()
  nome = input("Novo nome: ")
  idade = input("Nova idade: ")
  linhas[indice] = f"{nome},{idade}\n"
  with open(ARQUIVO, "w") as f:
    f.writelines(linhas)

def deletar():
    listar()
    indice = int(input("Digite o indice: "))
    with open(ARQUIVO, "r") as f:
      linhas = f.readlines()
    linhas.pop(indice)
    with open(ARQUIVO, "w") as f:
      f.writelines(linhas)

def menu():
    while True:
        print("1 - Adicionar")
        print("2 - Listar")
        print("3 - Buscar")
        print("4 - Atualizar")
        print("5 - Deletar")
        print("6 - Sair")
        op = input("Escolha: ")
        if op == "1": adicionar()
        elif op == "2": listar()
        elif op == "3": buscar()
        elif op == "4": atualizar()
        elif op == "5": deletar()
        elif op == "6": break
        else: print("Opção inválida")
menu()
```
